```python
def es_mayor(edad):
    return True if edad >= 18 else False

edades = [15, 22, 17, 30, 19, 12]
personas = ["Sofia", "Regina", "Micolete", "MIlena", "Jorge", "Luis"]

for i in range(len(edades)):
    if es_mayor(edades[i]):
        print(f"{personas[i]}, {edades[i]} años: pasa")
    else:
        print(f"{personas[i]}, {edades[i]} años: no pasa")
        
```
