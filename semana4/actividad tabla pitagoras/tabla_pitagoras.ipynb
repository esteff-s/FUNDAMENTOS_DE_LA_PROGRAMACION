Este programa construye una tabla de Pitágoras de 10 × 10 en una matriz con dos `for` enlazados, la imprime sin corchetes ni comas y permite consultar un producto buscándolo en la tabla, validando que los factores estén entre 1 y 10. Esta funcionando como una tipo calculadora pero de multiplicación sencilla

```python
#Tabla de pitagoras
# Parte A
tabla = []
for fila in range(1, 11):
    renglon = []
    for columna in range(1, 11):
        renglon.append(fila * columna)
    tabla.append(renglon)

# Parte B
def imprimir_tabla(tabla):
    for renglon in tabla:
        for numero in renglon:
            print(numero, end="\t")
        print()  

# Parte C
def consultar_producto(tabla, renglon, columna):
    return tabla[renglon - 1][columna - 1]

# Parte D
imprimir_tabla(tabla)

renglon = int(input("Renglón (1-10): "))
while renglon < 1 or renglon > 10:
    print("Fuera de rango: debe estar entre 1 y 10.")
    renglon = int(input("Renglón (1-10): "))

columna = int(input("Columna (1-10): "))
while columna < 1 or columna > 10:
    print("Fuera de rango: debe estar entre 1 y 10.")
    columna = int(input("Columna (1-10): "))

resultado = consultar_producto(tabla, renglon, columna)
print(f"{renglon} x {columna} = {resultado}")
```
