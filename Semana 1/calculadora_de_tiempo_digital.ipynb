```python
nombre = input("¿Cómo te llamas? ")
horas_redes = float(input("Horas al día en redes sociales: "))
horas_mensajeria = float(input("Horas al día en mensajería: "))
horas_streaming = float(input("Horas al día en streaming (series, videos): "))
horas_libros = float(input("Horas al día en libros: "))
horas_videos = float(input("Horas al día en videos: "))
horas_Spotify = float(input("Horas al día en Spotify: "))


tiempo_total = horas_redes + horas_mensajeria + horas_streaming + horas_libros + horas_videos
porcentaje = (tiempo_total / 24) * 100

print("----- RESUMEN DE TU RUTINA DIGITAL -----")
print(f"Usuario: {nombre}")
print(f"Tiempo total en pantalla: {tiempo_total} horas")
print(f"Porcentaje de tu día: {porcentaje:.2f}%") 
```
