# Las reglas del museo
Precio base según la edad:

Edad	Precio base

Menos de 3 años	Gratis ($0)

De 3 a 17 años	$30

18 años o más	$45

Descuentos (solo para personas de 18 años o más):

Tipo de visitante	Cómo se identifica	Descuento

Adulto mayor	Automático si tiene 60 años o más	12%

Estudiante	Lo elige en el menú (opción 1)	10%

Profesor	Lo elige en el menú (opción 2)	10%

Ninguno	Lo elige en el menú (opción 3)	0%

```python
# Precios
PRECIO_MENOR_3 = 0
PRECIO_MENOR = 30
PRECIO_ADULTO = 45

# Descuentos
DESC_ADULTO_MAYOR = 0.12
DESC_ESTUDIANTE = 0.10
DESC_PROFESOR = 0.10
DESC_NINGUNO = 0
```

```python
def taquilla():
    # Acumuladores y contador
    visitantes_cobrados = 0
    total_descuentos = 0
    total_a_pagar = 0

    print("=== TAQUILLA DEL MUSEO ===")

    while True:
        entrada = input("Edad del visitante (o 'fin' para cerrar la taquilla): ")

        if entrada == "fin":
            break

        if not entrada.isdigit():
            print("Edad no válida, no se cobra.")
            continue

        edad = int(entrada)

        if edad < 3:
            print(f"Edad: {edad} | Menor de 3 años | Entrada gratis | A pagar: ${PRECIO_MENOR_3:.2f}")
            continue

        if edad <= 17:
            tipo = "Menor"
            base = PRECIO_MENOR
            porcentaje = 0
        else:
            base = PRECIO_ADULTO
            if edad >= 60:
                tipo = "Adulto mayor"
                porcentaje = DESC_ADULTO_MAYOR
            else:
                opcion = input("Tipo: 1) Estudiante  2) Profesor  3) Ninguno: ")
                if opcion == "1":
                    tipo = "Estudiante"
                    porcentaje = DESC_ESTUDIANTE
                elif opcion == "2":
                    tipo = "Profesor"
                    porcentaje = DESC_PROFESOR
                else:
                    tipo = "Ninguno"
                    porcentaje = DESC_NINGUNO

        descuento = base * porcentaje
        pagar = base - descuento

        visitantes_cobrados += 1
        total_descuentos += descuento
        total_a_pagar += pagar

        print(f"Visitante {visitantes_cobrados} | Edad: {edad} | {tipo} | "
              f"Base: ${base:.2f} | Descuento: ${descuento:.2f} | A pagar: ${pagar:.2f}")

    print("=== CORTE DE CAJA ===")
    print(f"Visitantes cobrados: {visitantes_cobrados}")
    print(f"Total en descuentos: ${total_descuentos:.2f}")
    print(f"TOTAL A PAGAR: ${total_a_pagar:.2f}")
    return total_a_pagar
```
