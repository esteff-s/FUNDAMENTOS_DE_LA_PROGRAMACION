# Las reglas del museo
Precio base según la edad:

Edad	Precio base

Menos de 3 años	Gratis ($0)

De 3 a 17 años	$30

18 años o más	$45

Descuentos (solo para personas de 18 años o más):

Tipo de visitante	Cómo se identifica	Descuento

Adulto mayor	Automático si tiene 60 años o más	12%

Estudiante	Lo elige en el menú (opción 1)	10%

Profesor	Lo elige en el menú (opción 2)	10%

Ninguno	Lo elige en el menú (opción 3)	0%

```python
 #PRECIO visitantes por edad  
Niños = 0
Menores = 30
Adultos = 45

# Descuentos

A_Mayor = 0.12
Estudiante = 0.10
Profesor = 0.10
Ninguno = 0

funcionamiento = True

while funcionamiento == True:
        edad = int(input("Escribe tu edad: "))

if edad < 3:
     print(f"El visitante no paga boleto (niños): ${Niños}")
     funcionamiento = False
else: 
     print("Que tipo de visitante eres")
     print("1. Estudiante")
     print("2. Profesor")
     print("2. Adulto mayor")
     print("3. Ninguno") 
     print ()
     tipo = int(input("Introduce un valor del 1 al 4: "))


if edad <= 3 and edad >= 17:
        print(f"El visitante paga boleto de menor: ${Menores}")
        precio_entrada_menor = Menores 
        if tipo == 1:
            print(f"El visitante es estudiante y tiene un descuento del 10%: ${Menores * Estudiante}")
            precio_entrada_menor = Menores - (Menores * Estudiante)
        elif tipo == 4:
            print(f"El visitante no tiene descuento: ${Menores * Ninguno}")
            precio_entrada_menor = Menores - (Menores * Ninguno)     

elif edad >= 18:
        print("El visitante paga boleto de adulto")
        precio_entrada_adulto = Adultos 
        if tipo == 2:
            print(f"El visitante es profesor y tiene un descuento del 10%: ${Adultos * Profesor}")
            precio_entrada_adulto = Adultos - (Adultos * Profesor)
        elif tipo == 3:
                print(f"El visitante es adulto mayor y tiene un descuento del 12%: ${Adultos * A_Mayor}")
                precio_entrada_adulto = Adultos - (Adultos * A_Mayor)
        else tipo == 4:
            print(f"El visitante no tiene descuento: ${Adultos * Ninguno}")
            precio_entrada_adulto = Adultos - (Adultos * Ninguno)







    

```
