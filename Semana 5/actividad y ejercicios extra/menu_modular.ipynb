Este programa es para un menú que funciona como la consola de un concierto equis: cada opción hace una función que demuestra tuplas, diccionarios, excepciones o strings. Después de seleccionar cada sección el menú vuelve a aparecer y solo termina al elegir la opción 5.

```python
# PARTE A
numeros = (4, 9, 2, 7, 5, 1)          # tupla con al menos 5 números


def pedir_entero(mensaje):
    """Pide un entero y vuelve a preguntar hasta que el dato sea válido."""
    while True:
        try:
            return int(input(mensaje))
        except ValueError:
            print("Dato inválido, escribe un número entero.")


def sumar_tupla(tupla):
    """Suma todos los elementos de la tupla con un for y regresa el total."""
    total = 0
    for n in tupla:
        total = total + n
    return total


def seccion_tuplas():
    print("\n--- TUPLAS ---")
    print("Tupla original:", numeros)
    print("Tercer elemento:", numeros[2])

    a = pedir_entero("Primer número: ")
    b = pedir_entero("Segundo número: ")

    nuevas = numeros + (a, b)          # tupla NUEVA; la original no cambia
    print("Nueva tupla:", nuevas)
    print("La original sigue igual:", numeros)

    lista = list(nuevas)               # copia a lista para poder ordenar
    lista.sort()
    print("Lista ordenada:", lista)

    print("Suma de la nueva tupla:", sumar_tupla(nuevas))

# PARTE B

contactos = {"Ana": "555-0101", "Luis": "555-0102", "Mía": "555-0103"}


def buscar_telefono(diccionario, nombre):
    """Regresa el teléfono del nombre, o None si no existe (sin KeyError)."""
    if nombre in diccionario:
        return diccionario[nombre]
    return None


def seccion_diccionarios():
    print("\n--- DICCIONARIOS ---")

    nombre = input("Nombre del nuevo contacto: ").strip().title()
    telefono = input("Teléfono: ").strip()
    contactos[nombre] = telefono
    print(f"Se agregó a {nombre}.")

    print("Contactos registrados:")
    for n in contactos.keys():         # solo las claves (los nombres)
        print(" -", n)

    buscado = input("¿A quién buscas? ").strip().title()
    resultado = buscar_telefono(contactos, buscado)
    if resultado is None:
        print(f"{buscado} no está en tus contactos")
    else:
        print(f"Teléfono de {buscado}: {resultado}")

# PARTE C

def seccion_excepciones():
    print("\n--- EXCEPCIONES ---")
    try:
        a = int(input("Primer número entero: "))
        b = int(input("Segundo número entero: "))
        suma = a + b
        division = a / b               # aquí puede ocurrir ZeroDivisionError
        print(f"Suma: {suma}")
        print(f"División: {division}")
    except ValueError:
        print("Eso no es un número entero. Escribe solo dígitos (sin letras ni dejarlo vacío).")
    except ZeroDivisionError:
        print("No se puede dividir entre cero. Prueba con otro divisor.")

# PARTE D

def contar_palabras(texto):
    """Regresa cuántas palabras tiene el texto."""
    return len(texto.split())


def seccion_strings():
    print("\n--- STRINGS ---")
    mensaje = "el metal nunca muere en vivo"   # más de 4 palabras

    print("Mensaje:", mensaje)
    print("Longitud:", len(mensaje))
    print("En mayúsculas:", mensaje.upper())
    print("Con reemplazo:", mensaje.replace("metal", "rock"))
    print("Número de palabras:", contar_palabras(mensaje))

# PARTE E

def menu():
    while True:
        print("\n1) Tuplas  2) Diccionarios  3) Excepciones  4) Strings  5) Salir")
        opcion = input("Elige una opción: ")   # se compara como texto

        if opcion == "1":
            seccion_tuplas()
        elif opcion == "2":
            seccion_diccionarios()
        elif opcion == "3":
            seccion_excepciones()
        elif opcion == "4":
            seccion_strings()
        elif opcion == "5":
            print("Fin del programa. ¡Gracias!")
            break
        else:
            print("Opción inválida")


menu()
```
