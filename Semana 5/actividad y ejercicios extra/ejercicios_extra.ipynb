```python
#Ejercicio 1 

dias = ("Lunes", "Martes", "Miércoles", "Jueves", "Viernes", "Sábado", "Domingo")
km = (120, 85, 0, 210, 95, 0, 150)

def resumen(km):
    total = 0
    rodados = 0
    idx_mayor = 0
    for i in range(len(km)):
        total += km[i]
        if km[i] > 0:
            rodados += 1
        if km[i] > km[idx_mayor]:
            idx_mayor = i
    promedio = round(total / rodados, 1)
    return (total, rodados, promedio, idx_mayor)

total, rodados, promedio, idx = resumen(km)
print(f"Total: {total} km")
print(f"Días con rodada: {rodados}")
print(f"Promedio por día rodado: {promedio} km")
print(f"Día más largo: {dias[idx]} ({km[idx]} km)")
```

```python
#Ejercicio 2 
votos = [("Metallica", 48), ("Slayer", 22), ("Megadeth", 35), ("Anthrax", 15), ("Iron Maiden", 40)]

def promedio_votos(lista):
    suma = 0
    for banda, v in lista:
        suma += v
    return suma / len(lista)

# Ganador sin max() ni sort()
ganador, mayor = votos[0]
for banda, v in votos:
    if v > mayor:
        ganador, mayor = banda, v

prom = promedio_votos(votos)
print(f"Ganador: {ganador} con {mayor} votos")
print(f"Promedio: {prom}")
print("Arriba del promedio:")
for banda, v in votos:
    if v > prom:
        print(f"- {banda} ({v})")
```

```python
#Ejercicio 3 
paradas = (("Querétaro", 0), ("San Juan del Río", 52), ("Tequisquiapan", 74), ("Bernal", 125), ("Jalpan", 260))

mayor = 0
tramo_mayor = ""
for i in range(len(paradas) - 1):
    nombre1, km1 = paradas[i]
    nombre2, km2 = paradas[i + 1]
    distancia = km2 - km1
    print(f"{nombre1} → {nombre2}: {distancia} km")
    if distancia > mayor:
        mayor = distancia
        tramo_mayor = f"{nombre1} → {nombre2}"

print(f"Tramo más largo: {tramo_mayor} ({mayor} km)")
```

```python
#Ejercicio 4
temps = (8, 14, 22, 27, 31, 19, 11)

def clasificar(t):
    if t < 12:
        return "frío"
    elif t < 25:
        return "templado"
    else:
        return "calor"

etiquetas = []
contador = {}
for t in temps:
    et = clasificar(t)
    etiquetas.append(et)
    if et in contador:
        contador[et] += 1
    else:
        contador[et] = 1

print(etiquetas)
print(contador)

predomina = ""
mayor = 0
for et, n in contador.items():
    if n > mayor:
        mayor = n
        predomina = et
print(f"Predomina: {predomina}")
```

```python
#Ejercicio 5
setlist = (("Intro", 95), ("Riff asesino", 245), ("Balada", 302), ("Solo", 128), ("Encore", 409))

def formato(seg):
    minutos = seg // 60
    segundos = seg % 60
    if segundos < 10:
        return f"{minutos}:0{segundos}"
    return f"{minutos}:{segundos}"

total = 0
for cancion, seg in setlist:
    print(f"{cancion}: {formato(seg)}")
    total += seg
print(f"Duración total: {formato(total)}")
```

```python
#Ejercicio 6
inventario = {"bujía": 10, "cadena": 2, "balata": 6, "filtro": 4}
pedidos = [("bujía", 3), ("cadena", 5), ("espejo", 1), ("balata", 6), ("filtro", 1)]

for pieza, cantidad in pedidos:
    if pieza not in inventario:
        print(f"{pieza}: no existe en el taller")
    elif inventario[pieza] < cantidad:
        print(f"{pieza}: stock insuficiente (hay {inventario[pieza]})")
    else:
        inventario[pieza] -= cantidad
        print(f"{pieza}: surtido ({cantidad})")

print(inventario)
agotados = []
for pieza, cantidad in inventario.items():
    if cantidad == 0:
        agotados.append(pieza)
print(f"Agotados: {agotados}")
```

```python
#Ejercicio 7 
notas = {"Ana": [90, 85, 78], "Luis": [60, 72, 65], "Mía": [100, 95, 98], "Beto": [70, 68, 74]}

def promedio(lista):
    suma = 0
    for n in lista:
        suma += n
    return round(suma / len(lista), 1)

promedios = {}
for alumno, lista in notas.items():
    promedios[alumno] = promedio(lista)

mejor = ""
mayor = 0
for alumno, prom in promedios.items():
    estado = "Aprobado" if prom >= 70 else "Reprobado"
    print(f"{alumno}: {prom} - {estado}")
    if prom > mayor:
        mayor = prom
        mejor = alumno
print(f"Mejor promedio: {mejor}")
```

```python
#Ejercicio 9
precios = {"playera": 350, "gorra": 220, "parche": 80, "sudadera": 650}
carrito = ["playera", "gorra", "playera", "parche", "parche", "parche"]

cuenta = {}
for producto in carrito:
    if producto in cuenta:
        cuenta[producto] += 1
    else:
        cuenta[producto] = 1

subtotal = 0
for producto, piezas in cuenta.items():
    importe = piezas * precios[producto]
    print(f"{piezas} x {producto} = ${importe}")
    subtotal += importe

descuento = 0
if subtotal > 800:
    descuento = subtotal * 0.10

print(f"Subtotal: ${subtotal}")
print(f"Descuento: ${descuento}")
print(f"Total: ${subtotal - descuento}")
```

```python
#Ejercicio 10
abrev = {"q": "que", "xq": "porque", "tmb": "también", "pq": "porque", "bn": "bien", "tqm": "te quiero mucho"}

def expandir(mensaje):
    palabras = mensaje.lower().split()
    resultado = []
    for p in palabras:
        resultado.append(abrev.get(p, p))
    return " ".join(resultado)

print(expandir("Q onda tmb voy al concierto xq tocan bn"))
print(expandir("tqm Mamá"))
```

```python
#Ejercicio 11
# Entradas de prueba: trescientos, 300, 0, 300, 16
def rendimiento():
    while True:
        try:
            km = float(input("Km recorridos: "))
            litros = float(input("Litros cargados: "))
            resultado = round(km / litros, 1)
            return resultado
        except ValueError:
            print("Escribe solo números")
        except ZeroDivisionError:
            print("Los litros no pueden ser 0")

r = rendimiento()
print(f"Rendimiento: {r} km/l")
if r >= 20:
    print("Excelente")
elif r >= 15:
    print("Normal")
else:
    print("Revisa tu moto")
```

```python
#Ejercicios 12
# Entradas de prueba: diez, 10, -5, 35, 70
def precio_boleto(edad):
    if edad < 12:
        return 150
    elif edad < 60:
        return 300
    else:
        return 180

total = 0
for persona in range(3):
    while True:
        try:
            edad = int(input(f"Edad de la persona {persona + 1}: "))
            if edad < 0 or edad > 120:
                print("Edad fuera de rango")
            else:
                break
        except ValueError:
            print("Escribe la edad con número")
    precio = precio_boleto(edad)
    print(f"Boleto: ${precio}")
    total += precio

print(f"Total a pagar: ${total}")
```

```python
#Ejercicio 13
capturas = ["10", "7", "x", "", "9", "8.5", " 6 "]

validas = []
invalidas = 0
for c in capturas:
    try:
        validas.append(int(c))
    except ValueError:
        invalidas += 1

print(f"Válidas: {validas}")
print(f"Inválidas: {invalidas}")

suma = 0
for n in validas:
    suma += n
print(f"Promedio: {round(suma / len(validas), 2)}")
```

```python
#Ejercicio 14
# Entradas de prueba: uno, 0, 5, 0, 0, 1, 2
sala = [["X", "L", "L"], ["L", "X", "L"], ["X", "X", "L"]]

def reservar(sala):
    try:
        fila = int(input("Fila: "))
        col = int(input("Columna: "))
        if sala[fila][col] == "X":
            print("Ocupado, elige otro")
            return False
        sala[fila][col] = "X"
        print("¡Reservado!")
        return True
    except ValueError:
        print("Escribe números")
        return False
    except IndexError:
        print("Ese asiento no existe")
        return False

reservado = False
while not reservado:
    reservado = reservar(sala)

for fila in sala:
    for asiento in fila:
        print(asiento, end=" ")
    print()
```
